```python
#3
scores = np.array([
#quiz,midterm,final
[18, 15, 20],
[12, 14, 16],
[20, 19, 18],
[10, 8, 15]
]) #(4,3)
scheme_A = np.array([0.5, 0.3, 0.2]) #0
scheme_B = np.array([0.2, 0.3, 0.5]) #1
scheme_C = np.array([0.1, 0.2, 0.7]) #2
```

```python
#3-1
# weight_matrix (3,3)
weight_matrix=np.array([scheme_A,scheme_B,scheme_C])
print(f"weight_matrix :\n {weight_matrix}")
print(weight_matrix.shape)
```

```python
#3-2 @
final_scores = scores @ weight_matrix.T
print(final_scores)
print(final_scores.shape)
```

```python
#3-3 np.argmax axis=1
max_scores=np.argmax(final_scores,axis=1)
print(max_scores)
max_scores=np.max(final_scores,axis=1)
print(max_scores)
#A-0
#B-1
#C-2
```

```python
#3-4
print(f"final_scores : \n {final_scores}")

mean_scores = np.mean(final_scores,axis=0)#baraye har sutun yani har ravesh
print(mean_scores)
best=np.argmax(mean_scores)
print(best)
print(np.array(mean_scores)[best],"--> A")
```

```python
#6-1
bib_numbers = np.array([101, 102, 103, 104, 105, 106])
times_5k = np.array([22.3, 25.1, 21.8, 26.4, 23.0, 24.7])


bib_numbers_2 = np.array([107, 108])
times_5k_2 = np.array([20.5, 27.9])

bib_numbers = np.concatenate((bib_numbers,bib_numbers_2))
times_5k = np.concatenate((times_5k,times_5k_2))
print(bib_numbers)
print(times_5k)
```
