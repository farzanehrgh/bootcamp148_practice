```python
#2
import numpy as np

import csv

lst=[]
with open ("cities.csv","r") as file:
    read = csv.DictReader(file)
    for i in read:
        try:
            i["humidity"]=int(i["humidity"])
            lst.append(i)
        except:
            print("warning")
            continue
print(lst)
```

```python
city_names=[]

for i in lst:
    city_names.append(i["city"])
print("city_names : \n",city_names)


# temp=[]
# for i in lst:
#     temp.append(i["avg_temp"])
#     print(temp)

data=[]
for i in lst:
    row=[int(i["avg_temp"]),int(i["humidity"]),int(i["popularity"]),int(i["cost_index"])]
    data.append(row)

data=np.array(data)
print("data : \n",data)
print(data.shape)
print(data.ndim)
print(data.size)
print(data.dtype)
```

```python
import numpy as np

means=np.mean(data,axis=0)
print(means)

mins=np.min(data,axis=0)
print(mins)

maxi=np.max(data,axis=0)
print(maxi)

std=np.std(data,axis=0)
print(std)
```

max_min scaling: furmula : xscaled=x-xmin/xmax-xmin
meghyas bandie dada ha be baze 0,1


```python
scaled_data=(data-mins)/(maxi-mins)
print(scaled_data)
```

```python
ideal = np.array([26, 40, 8, 4])
print(ideal)

scaled_ideal=(ideal-mins)/(maxi-mins)
print(scaled_ideal)
```

```python
d=np.linalg.norm(scaled_data-scaled_ideal,axis=1)
print(d)
min_city=(np.argmin(d))
print(min_city)
print(np.min(d))
print(city_names[min_city],min_city)

max_city=np.argmax(d)
print(max_city)
print(city_names[max_city],max_city)
```

```python
cosine_similarity=np.dot(scaled_data,scaled_ideal)/(np.linalg.norm(scaled_data,axis=1)*(np.linalg.norm(scaled_ideal)))
print(cosine_similarity)

cs=np.argmax(cosine_similarity)
print(city_names[cs],cs)
print(city_names[np.argmax(cosine_similarity)], "with a similarity of", np.max(cosine_similarity))
```

```python
weights = np.array([0.1, 0.2, 0.3, 0.4])
w_data=scaled_data@weights
print(weights)
print(w_data)
w_ideal=scaled_ideal@weights
print(w_ideal)
```

```python
temps = np.array([
[22, 24, 19, 21], 
[25, 27, 23, 24],
[18, 20, 17, 19] 
])
#axis=1
monday_avg_temp=np.mean(temps[0,:])
tuesday_avg_temp=np.mean(temps[1,:])
wednesday_avg_temp=np.mean(temps[2,:])
print(monday_avg_temp)
print(tuesday_avg_temp)
print(wednesday_avg_temp)
#axis0
first_column_avg=np.mean(temps[:,0])
second_column_avg=np.mean(temps[:,1])
third_column_avg=np.mean(temps[:,2])
forth_column_avg=np.mean(temps[:,3])
print(first_column_avg)
print(second_column_avg)
print(third_column_avg)
print(forth_column_avg)
```

```python
#broadcasting

daily_mean=np.mean(temps,axis=1)
print(daily_mean)
print(temps.shape)
print(daily_mean.shape)
diff=temps-daily_mean[:,np.newaxis]
print(diff)

reshape=daily_mean.reshape(3,1)
print(reshape)
diff1=temps-reshape
print(diff1)


```
