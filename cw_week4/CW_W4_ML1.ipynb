 Warm-up: Dataset, Sample, Feature, Label

```python
#1
import numpy as np

# study_hours, attendance_percent, previous_score, final_score
students = np.array([
 [2, 60, 45, 50],
 [5, 80, 70, 75],
 [1, 40, 35, 38],
 [8, 90, 88, 92],
 [4, 75, 60, 65],
 [7, 85, 80, 84],
 [3, 55, 50, 52],
 [6, 70, 72, 78]
])
#1
print(students.dtype)
print(students.size)
print(students.shape)
print(students.ndim)
#2 

# --->8 students -->8 samples in dataset
samples=len(students)
print(samples)

# --->4 features in dataset
features=len(students[0,:])
print(features)

#4
x= students[:,:3]
print(x)
print(x.shape)

#5
y=students[:,-1]
print(y)
print(y.shape)


```

Machine Learning یا — Rule-based AI

```python
#2
print(y)
rule_based_label=np.where(y>=60,"pass","fail")
print(rule_based_label)
print(np.sum(rule_based_label=="pass"))
print(np.sum(rule_based_label=="fail"))

```

 Feature Engineering

```python
#3. new feature
# study_hours, attendance_percent, previous_score
print(x)
#1
study_hours = x[:,0]
print(study_hours)
mean=np.mean(x[:,0])
print(mean)

#max,min
attendance_percent = x[:,1]
print(attendance_percent)
maxx=np.max(x[:,1])
minn=np.min(x[:,1])
print(maxx)
print(minn)

#4
effort_score = study_hours * attendance_percent / 100
print(effort_score)
#5
x_new=np.column_stack((x,effort_score))
print(x_new)

#6
print(x_new.shape)

# effort_score.reshape(-1, 1)
# np.concatenate(...)
# or
# np.column_stack(...)
```

Classification یا — Regression 

```python
#2 labels pas/fail

print(y)
class_label=np.where(y>=60,"pass","fail")
print(class_label)
passed_students=np.sum(class_label=="pass")
print(passed_students)
passed_percentage=passed_students/len(y)*100
print(passed_percentage)

```

 Multi-class Label

```python
#3 labels weak /normal/strong

label_1=np.where(y>=80,"strong",y)
label_2=np.where(y<80,"normal",y)
label_3=np.where(y<60,"weak",y)
print("label1",label_1)
print("label2",label_2)
print("label3",label_3)
print(np.sum(label_1=="strong"))
print(np.sum(label_2=="normal"))
print(np.sum(label_3=="weak"))
strong_mean=np.mean(study_hours[label_1=="strong"])
print(strong_mean)
normal_mean=np.mean(study_hours[label_2=="normal"])
print(normal_mean)
weak_mean=np.mean(study_hours[label_3=="weak"])
print(weak_mean)

```
